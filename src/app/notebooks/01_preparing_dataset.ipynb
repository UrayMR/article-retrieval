```python
# Load Library

import numpy as np
import pandas as pd
from pathlib import Path
from lingua import Language, LanguageDetectorBuilder

DATA_FOLDER_FILE = Path('../../data/')
print(f"Data folder path: {DATA_FOLDER_FILE}")
```

```python
# Load the dataset
INPUT_PATH = Path(DATA_FOLDER_FILE / 'raw/main_data.csv')

df = pd.read_csv(INPUT_PATH, low_memory=False)
df.head()
```

```python
df.info()
```

```python
df[['Title']].sample(20, random_state=42)
```

# Filter the dataset by removing rows with missing values in the Selected columns


```python
# Filter the dataset by removing rows with missing values in the 'ID', 'Title' and 'Abstrak' columns
selected_columns = ['ID', 'Title', 'Abstrak']

df_filtered = df.dropna(subset=selected_columns)

df_filtered.info()
```

```python
# Lowercase all the selected columns
df_filtered[selected_columns] = df_filtered[selected_columns].apply(lambda x: x.str.lower())

df_filtered[selected_columns].head()
```

# Detect Language of the Title Columns


```python
df_selected = df_filtered[selected_columns]

df_selected.info()
```

```python
detector = LanguageDetectorBuilder.from_languages(
    Language.INDONESIAN,
    Language.ENGLISH
).build()

def detect_language(text):
    if pd.isna(text) or not str(text).strip():
        return pd.Series({
            "language": "missing",
            "confidence": np.nan,
        })

    text = str(text).strip()
    try:
        confidence_values = detector.compute_language_confidence_values(text)
        detected = detector.detect_language_of(text)

        if detected is None or len(confidence_values) == 0:
            return pd.Series({
                "language": "uncertain",
                "confidence": np.nan,
            })

        language = detected.iso_code_639_1.name.lower()
        top_confidence = float(confidence_values[0].value)

        return pd.Series({
            "language": language,
            "confidence": top_confidence,
        })
    except Exception as exc:
        return pd.Series({
            "language": "uncertain",
            "confidence": np.nan,
        })
```

```python
# Deteksi judul dan abstrak secara terpisah; teks asli dipertahankan
title_lang = df_selected["Title"].apply(detect_language).add_prefix("title_")
abstract_lang = df_selected["Abstrak"].apply(detect_language).add_prefix("abstract_")

df_audit = pd.concat([df_selected, title_lang, abstract_lang], axis=1)

display(df_audit[[
    "ID", "Title", "Abstrak",
    "title_language", "title_confidence",
    "abstract_language", "abstract_confidence",
]].sample(min(20, len(df_audit)), random_state=42))
```

# Translate Title Column to Indonesian


```python
# # Add column for translated content
# e.g. =GOOGLETRANSLATE(A1, "en", "id")
TRANSLATE_TEXT = "=GOOGLETRANSLATE({}; \"en\"; \"id\")"

# Only translate titles that have _language = "en"
# if title_language is not "en", then leave it as is (no translation needed), copy the original title to the new column
title_english_mask = title_lang["title_language"].eq("en")
abstract_english_mask = abstract_lang["abstract_language"].eq("en")

df_selected["title_translated"] = pd.Series(index=df_selected.index, dtype="object")
df_selected.loc[title_english_mask, "title_translated"] = df_selected.loc[title_english_mask, "Title"].astype(str).apply(
    lambda x: TRANSLATE_TEXT.format(f'"{x}"')
)

df_selected["abstract_translated"] = pd.Series(index=df_selected.index, dtype="object")
df_selected.loc[abstract_english_mask, "abstract_translated"] = df_selected.loc[abstract_english_mask, "Abstrak"].astype(str).apply(
    lambda x: TRANSLATE_TEXT.format(f'"{x}"')
)

df_selected.head()
```

```python
OUTPUT_FOLDER = Path(DATA_FOLDER_FILE / 'translated')
OUTPUT_FOLDER.mkdir(parents=True, exist_ok=True)

df_selected.to_csv(OUTPUT_FOLDER / 'selected_data.csv', index=False)

print(f"Selected data saved to: {OUTPUT_FOLDER / 'selected_data.csv'}")
```
